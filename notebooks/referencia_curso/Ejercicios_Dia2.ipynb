## Ejercicio 2.1 · Un patrón que falta

En clase vimos que las expresiones regulares detectan los documentos peruanos que las
herramientas comerciales no conocen. Pero nuestra lista estaba incompleta.

Escribe un patrón que detecte un **carné de extranjería**, que en el Perú tiene nueve dígitos
seguidos.

```python
import re

TEXTO = ("Soy Kenji Nakamura Flores, carné de extranjería 001234567, "
         "mi DNI anterior era 45871203 y mi teléfono 987654321.")

# ESCRIBE TU PATRÓN AQUÍ
# Pista: \b marca el borde de una palabra y \d{n} significa n dígitos seguidos.

patron_carne = r""       # <-- complétalo

encontrados = re.findall(patron_carne, TEXTO) if patron_carne else []
print("encontrados:", encontrados)
```

```python
comprobar("2.1",
          "001234567" in encontrados,
          "Detectaste el carné de extranjería.",
          "Debe encontrar 001234567. Prueba con nueve dígitos entre bordes de palabra.")

# Una segunda comprobación, más interesante que la primera:
falsos = [e for e in encontrados if e in ("45871203", "987654321")]
print()
comprobar("2.1 bis",
          len(falsos) == 0,
          "Y además no confundió el DNI ni el teléfono, que tienen otra longitud.",
          f"Cuidado: tu patrón también atrapó {falsos}. Revisa los bordes de palabra.")
```
