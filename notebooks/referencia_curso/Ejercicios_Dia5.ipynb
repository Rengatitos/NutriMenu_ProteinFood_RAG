```python
mi_criterio = """
ESCRIBE AQUÍ TU CRITERIO
Ejemplo de la forma que debe tener:
"Se retira si, tras N meses en operación, [indicador medible] no supera [valor],
 medido con [fuente de datos]."
"""

print(mi_criterio)
```

No hay comprobación automática. Revisa tú mismo si tu criterio cumple las tres condiciones:

Tiene un **número**, no un adjetivo. "Si no mejora la atención" no sirve; "si la satisfacción no
supera el 70%" sí.

Tiene un **plazo**. Sin plazo, siempre se puede pedir un trimestre más.

Dice **de dónde sale el dato**. Si nadie está midiendo ese indicador hoy, el criterio no se puede
aplicar y hay que montarlo antes de lanzar.

Escribirlo antes del piloto es lo que evita que un sistema que no funciona se sostenga durante
